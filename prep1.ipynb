```python
gymitrainer_df = pd.read_csv("/home/mlbd-project/updated_data/gymitrainer.csv")
display(gymitrainer_df.head())

mapping_df = pd.read_csv("/home/mlbd-project/updated_data/mapping.csv")
display(mapping_df.head())
```

```python
# read files
#gymitrainer_df = pd.read_csv("/home/mlbd_project/m4-gogymi/gymitrainer.csv")
#mapping_df = pd.read_csv("/home/mlbd_project/m4-gogymi/mapping.csv")

# Threshold
threshold = 0.5

# filtering
filtered_mapping = mapping_df[mapping_df['confidence'] >= threshold]

# Get only matching gymitrainer records (inner join)
matched_gymitrainer = pd.merge(
    gymitrainer_df,
    filtered_mapping[['id', 'user_id']],
    on='id',
    how='inner'  
)

# Sort by user_id
matched_gymitrainer = matched_gymitrainer.sort_values(by='user_id')
display(matched_gymitrainer.head())
```

```python
# Count the number of conversations for each user_id
usage_counts = matched_gymitrainer['user_id'].value_counts().reset_index()

# Edit column names
usage_counts.columns = ['user_id', 'usage_count']

# sort by user_id
usage_counts = usage_counts.sort_values(by='user_id')

display(usage_counts)
```
